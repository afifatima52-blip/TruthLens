```python
label_counts = df["label"].value_counts()

total = label_counts.sum()

fig, ax = plt.subplots(figsize=(12, 8))

wedges, _ = ax.pie(
    label_counts,
    startangle=90,
    wedgeprops={
        "width": 0.38,
        "edgecolor": "white",
        "linewidth": 2
    }
)

legend_labels = [
    f"{label}: {count:,} ({count / total:.1%})"
    for label, count in label_counts.items()
]

ax.legend(
    wedges,
    legend_labels,
    title="Fact-Check Labels",
    loc="center left",
    bbox_to_anchor=(1, 0.5),
    frameon=False
)

ax.text(
    0, 0,
    f"Total\n{total:,}",
    ha="center",
    va="center",
    fontsize=18,
    fontweight="bold"
)

ax.set_title(
    "TruthLens - Fact Check Label Distribution",
    fontsize=17,
    fontweight="bold",
    pad=20
)

ax.set_aspect("equal")

plt.tight_layout()
plt.show()
```

```python
# Select positive review delays
positive_delays = df.loc[
    df["review_delay_days"] > 0,
    "review_delay_days"
].dropna()

# Calculate outlier information
q1 = positive_delays.quantile(0.25)
q3 = positive_delays.quantile(0.75)
iqr = q3 - q1

upper_limit = q3 + (1.5 * iqr)

outlier_count = (positive_delays > upper_limit).sum()

# Create figure
fig, ax = plt.subplots(figsize=(12, 7))

fig.patch.set_facecolor("#F4F7FB")
ax.set_facecolor("white")

# Draw boxplot
sns.boxplot(
    y=positive_delays,
    ax=ax,
    color="#36BFC1",
    width=0.35,
    showfliers=False,
    linewidth=2,
    medianprops={"color": "#E76F51", "linewidth": 3},
    whiskerprops={"linewidth": 2},
    capprops={"linewidth": 2}
)

# Logarithmic scale
ax.set_yscale("log")

# Titles
ax.set_title(
    "TruthLens | Positive Fact-Checking Delay Distribution",
    fontsize=17,
    fontweight="bold",
    pad=20
)

ax.set_ylabel("Review Delay (Days)", fontsize=12)
ax.set_xlabel("Positive Review Delays", fontsize=12)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Add information
ax.text(
    0.98, 0.95,
    f"Records: {len(positive_delays):,}\nOutliers: {outlier_count:,}",
    transform=ax.transAxes,
    ha="right",
    va="top",
    fontsize=11,
    bbox=dict(
        boxstyle="round,pad=0.6",
        facecolor="#E8F4F5",
        edgecolor="none"
    )
)

plt.tight_layout()
plt.show()
```

```python
import plotly.express as px

# Convert review delay to numeric
df["review_delay_days"] = pd.to_numeric(
    df["review_delay_days"],
    errors="coerce"
)

# Prepare language summary
language_summary = (
    df.groupby("language")
    .agg(
        total_claims=("claim", "count"),
        reviewed_records=("review_delay_days", "count"),
        median_delay=("review_delay_days", "median")
    )
    .reset_index()
)

# Keep languages with available delay information
language_summary = language_summary.dropna(
    subset=["median_delay"]
)

# Create interactive bubble chart
fig = px.scatter(
    language_summary,
    x="median_delay",
    y="total_claims",
    size="reviewed_records",
    color="language",
    hover_name="language",
    hover_data={
        "total_claims": True,
        "reviewed_records": True,
        "median_delay": ":.2f"
    },
    size_max=55,
    title="TruthLens | Language-wise Claim Volume & Review Delay",
    labels={
        "median_delay": "Median Review Delay (Days)",
        "total_claims": "Total Claims",
        "language": "Language",
        "reviewed_records": "Records with Known Delay"
    },
    template="plotly_white"
)

fig.update_layout(
    title_x=0.5,
    title_font_size=21,
    height=700,
    font=dict(family="Arial", size=13),
    legend_title="Languages"
)

fig.update_traces(
    marker=dict(
        opacity=0.8,
        line=dict(width=1, color="white")
    )
)

fig.show()
```

```python
# Claim aur review years
df["claim_year"] = df["claimDate"].dt.year
df["review_year"] = df["reviewDate"].dt.year

# Har claimant ke claims ki frequency
df["claimant_activity"] = (
    df.groupby("claimant")["claimant"]
    .transform("count")
)

# Numerical columns
correlation_data = df[
    [
        "review_delay_days",
        "claim_year",
        "review_year",
        "claimant_activity"
    ]
]

print(correlation_data.head())
```

```python
# Correlation calculate karna
corr_matrix = correlation_data.corr()

# Upper triangle hide karna
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Figure
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    center=0,
    square=True,
    linewidths=2,
    linecolor="white",
    cbar_kws={"shrink": 0.8},
    annot_kws={"size": 12, "weight": "bold"}
)

plt.title(
    "TruthLens — Feature Correlation Analysis",
    fontsize=19,
    fontweight="bold",
    pad=20
)

plt.xticks(rotation=25, ha="right", fontsize=10)
plt.yticks(rotation=0, fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Top 5 languages
top_languages = df["language"].value_counts().head(5).index

# Language aur label ke counts
radar_data = pd.crosstab(
    df[df["language"].isin(top_languages)]["language"],
    df[df["language"].isin(top_languages)]["label"]
)

print(radar_data)
```

```python
import numpy as np

# Har language ke labels ko percentage mein convert karna
radar_percent = radar_data.div(
    radar_data.sum(axis=1), axis=0
) * 100

# Categories
categories = radar_percent.columns.tolist()
num_vars = len(categories)

# Angles calculate karna
angles = np.linspace(
    0, 2 * np.pi, num_vars, endpoint=False
).tolist()

angles += angles[:1]

# Radar chart
fig, ax = plt.subplots(
    figsize=(12, 12),
    subplot_kw={"polar": True}
)

colors = plt.cm.tab10(np.linspace(0, 1, len(radar_percent)))

for (language, row), color in zip(
    radar_percent.iterrows(), colors
):
    values = row.tolist()
    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2.5,
        label=language.upper(),
        color=color
    )

    ax.fill(
        angles,
        values,
        color=color,
        alpha=0.08
    )

# Formatting
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=9)

ax.set_ylim(0, 100)
ax.set_yticks([20, 40, 60, 80, 100])
ax.set_yticklabels(["20%", "40%", "60%", "80%", "100%"])

ax.set_title(
    "TruthLens — Fact-Check Label Distribution by Language",
    fontsize=16,
    fontweight="bold",
    pad=30
)

ax.legend(
    title="Languages",
    loc="upper right",
    bbox_to_anchor=(1.25, 1.12)
)

plt.tight_layout()
plt.show()
```

```python
sankey_data = (
    df.groupby(["language", "label"])
    .size()
    .reset_index(name="count")
)

print(sankey_data.head(15))
```

```python
import plotly.graph_objects as go

# Unique languages and labels
languages = sorted(sankey_data["language"].unique())
labels = sorted(sankey_data["label"].unique())

# Node labels
all_nodes = languages + labels

# Node indexes
node_dict = {
    name: index
    for index, name in enumerate(all_nodes)
}

# Source, target and values
sources = sankey_data["language"].map(node_dict).tolist()
targets = sankey_data["label"].map(node_dict).tolist()
values = sankey_data["count"].tolist()

# Sankey diagram
fig = go.Figure(
    data=[
        go.Sankey(
            arrangement="snap",
            node=dict(
                pad=20,
                thickness=22,
                line=dict(
                    color="white",
                    width=1
                ),
                label=all_nodes,
                color="#34495E"
            ),
            link=dict(
                source=sources,
                target=targets,
                value=values,
                color="rgba(52, 152, 219, 0.30)"
            )
        )
    ]
)

fig.update_layout(
    title=dict(
        text="TruthLens — Global Language to Fact-Check Label Flow",
        x=0.5,
        font=dict(size=21)
    ),
    font=dict(
        size=12,
        family="Arial"
    ),
    width=1200,
    height=750,
    paper_bgcolor="white",
    plot_bgcolor="white"
)

fig.show(renderer="vscode")
```
